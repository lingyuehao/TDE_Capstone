## Files with fee/cost/charge-related columns

```python
FEE_KEYWORDS = ["fee", "charge", "cost", "rate", "amount", "price"]


def fee_related_columns(columns):
    return [c for c in columns if any(k in str(c).lower() for k in FEE_KEYWORDS)]


csv_files = [
    "Database report1788270370660_Data Import Table.csv",
    "Database report1788271505299_Invoice table.csv",
    "Database report1788285611531_trackingnumber Table.csv",
    "Database report1788285968903_trackingdetail Table.csv",
]

excel_files = [
    "Cooper FedEx vs UPS Rerate Scenario Cooper Trumbull 123022 V1.xlsx",
    "Cooper Surgical Sample FedEx Net Rates.xlsx",
    "Cooper Surgical Sample UPS Net Rates.xlsx",
]

print("Files containing fee/cost/charge-related columns:\n")

for f in csv_files:
    cols = pd.read_csv(f, encoding=ENC, nrows=0).columns
    matches = fee_related_columns(cols)
    if matches:
        print(f"- {f}")
        print(f"    matched columns: {matches}\n")

for f in excel_files:
    xls = pd.ExcelFile(f)
    for sheet in xls.sheet_names:
        cols = pd.read_excel(f, sheet_name=sheet, nrows=0).columns
        matches = fee_related_columns(cols)
        if matches:
            print(f"- {f} [sheet: {sheet}]")
            print(f"    matched columns: {matches}\n")
```

## Amount columns vs. charge-type/category columns

Within the files above, columns split into two kinds:

- **Numeric $ amount columns** — an actual dollar figure.
- **Type/category columns** — a text label describing *what kind* of charge it is.

`trackingdetail Table.csv` (one row per charge line item):
- `Charge Value` -> numeric $ amount
- `Charge Type` / `Charge Description` -> the charge category label (this is the real "charge type" field)

`trackingnumber Table.csv` (one row per shipment, charges already broken into columns):
- $ amount columns: `Access Charge`, `Advance`, `Discounts`, `Discount`, `Fuel Charge`, `List Rate`, `Amount`, `Spec Charge`, `Spec Charge 2`-`7`, `Total Charge`, `Invoice Amount (CSV)`, `Invoice Charge`, `Amount (SB1)`-`(SB5)`, `Charge`, `Total Amount`, `Final Amount`, `AP Amount`, `AR Amount`, `Tariff`
- Looks like an amount but isn't (text/status columns): `Accessorial Charges` (values like "Not Verified"), `Amount Summary` (values like "Charges are same")
- Loosely category-like but sparse/cryptic, not a clean taxonomy: `Category`, `Code (SB1)`-`(SB5)`, `Description (SB1)`-`(SB5)`, `GL Code`, `Service Type`, `Service`, `Freight Type`

`Invoice table.csv`:
- `Total Amount` -> numeric $ amount, no category column at this level.

So the real "what are all the charge categories" question is answered by `Charge Type` / `Charge Description` in `trackingdetail Table.csv`.

```python
detail_types = pd.read_csv(
    "Database report1788285968903_trackingdetail Table.csv",
    encoding=ENC,
    usecols=["Charge Type", "Charge Description"],
    low_memory=False,
)

charge_type_counts = detail_types["Charge Type"].value_counts(dropna=False)
charge_desc_counts = detail_types["Charge Description"].value_counts(dropna=False)

print(f"Unique Charge Type values: {charge_type_counts.shape[0]:,}")
print(f"Unique Charge Description values: {charge_desc_counts.shape[0]:,}")
print()
print("Top 40 Charge Type values by row count:")
print(charge_type_counts.head(40))

# Full lists (1,800+ / 2,900+ raw labels) saved for manual review when building
# a category-mapping dict to sort these into a smaller set of buckets.
charge_type_counts.to_csv("outputs/charge_type_value_counts.csv", header=["count"])
charge_desc_counts.to_csv("outputs/charge_description_value_counts.csv", header=["count"])
print("\nSaved full lists to outputs/charge_type_value_counts.csv and outputs/charge_description_value_counts.csv")
```
